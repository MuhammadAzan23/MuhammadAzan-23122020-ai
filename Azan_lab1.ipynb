```python
destination = input("Where do you want to travel in Pakistan? ").lower()
budget = int(input("What is your budget (PKR)? "))

if destination == "hunza":
    if budget >= 80000:
        print("You can book a luxury trip to Hunza! 🏔️")
    else:
        print("You can book a budget trip to Hunza! 🏔️")

elif destination == "murree":
    if budget >= 50000:
        print("You can book a comfortable trip to Murree! 🌲")
    else:
        print("You can book a budget trip to Murree! 🌲")

elif destination == "skardu":
    if budget >= 100000:
        print("You can book a trip to Skardu! 🏔️")
    else:
        print("Your budget is too low for Skardu.")

elif destination == "lahore":
    if budget >= 30000:
        print("You can enjoy a trip to Lahore! 🕌")
    else:
        print("You can plan a short trip to Lahore.")

else:
    print("Sorry, we don't have a package for this destination.")

print("Thank you for using our Pakistan Travel Planner! 🇵🇰")
```
